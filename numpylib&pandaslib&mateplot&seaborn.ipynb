```python
import numpy as np

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

iris = np.genfromtxt(
    url,
    delimiter=",",
    dtype=float,
    usecols=(0, 1, 2, 3)
)

print(iris.shape)
condition=(iris[ : , 2 ]>0.2) & (iris[ : , 0 ]<0.5)
iris[condition]
np.corrcoef(iris[:,0],iris[: ,2])[0,1]
iris[np.random.randint(150,size=20),np.random.randint(4,size=20)]=np.nan
print(np.isnan(iris[:,0]).sum())
print(np.where(np.isnan(iris[:,0])))
```

 **MATEPLOTLIB**

```python
import matplotlib.pyplot as plt

df=pd.read_csv('tmdb_movies_data.csv')
print('------------------------------info about the data=--------------------------')
print(df.info())
print('--------------------first 5 rows about the data-------------------')
print(df.head())
```

```python
df.info()
```

```python
bins_num=[0,2,4,6,8,10]
plt.xlabel('vote avarage')
plt.ylabel('frecuancy')
plt.title('distribution of the avarage vote')
plt.hist(df['vote_average'],facecolor='cyan',edgecolor='maroon');
plt.show()
```

```python
print(df['vote_average'].max())
print(df['vote_average'].min())
```

```python
# plt.plot(df['vote_count'])
plt.plot([1,2,3,4,5,5])
plt.plot([10,20,30,40])
```

```python
movies_count=pd.DataFrame(df.release_year.value_counts(ascending=True))
movies_count
```

```python
movies_count=movies_count.reset_index()
movies_count
movies_count.columns=['year','count']
```

```python
movies_count=movies_count.sort_values(by=['year'])
movies_count
```

```python
plt.plot(
    movies_count['year'],
    movies_count['count'],
    c='0.7',
    linestyle='-',
    marker='*'
)
plt.plot(range(1960,201))
plt.legend(['Movies', 'Series'], shadow=True, loc=1)
plt.xlabel('year')
plt.ylabel('count')
plt.title('release of every movie')
plt.xlim(1960,2015)
plt.xticks(range(1960,2015,5));
plt.axis('off')
```

```python
top_geners=df.groupby('genres').sum().sort_values(by='revenue',ascending=False)['revenue'][0:5]
top_geners
```

```python
plt.pie(top_geners.values,labels=top_geners.index,colors=['red','gray','blue','green','pink'],explode=[0.2,0,0,0,0],autopct='%1.2f%%');
```

```python
plt.scatter(df.vote_average,df.revenue,alpha=0.4,marker='+',s=100);
plt.scatter(df.vote_count,df.revenue,alpha=0.1,s=100);
plt.xlabel('vote avarage&count')
plt.ylabel('revenue')
```
